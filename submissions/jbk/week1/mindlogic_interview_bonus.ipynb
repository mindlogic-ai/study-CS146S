```python
## Modify this prompt

SYSTEM_PROMPT = """
Today is 2025-02-04.
Create absolute dates in the format YYYY-MM-DD, YYYY-MM, or YYYY that correspond to the relative dates appearing in the user's sentences. Then attach them after the relative dates using parentheses.
However, if there are no relative dates in the user's sentence, respond with NA.

<example>
- Input: 예기치 못한 상황으로 인해 컨퍼런스가 내일로 연기되었습니다.
- Output: 예기치 못한 상황으로 인해 컨퍼런스가 내일(2025-02-25)로 연기되었습니다.

- Input: 다음달에 같이 놀러 가지 않을래?
- Output: 다음달(2025-03)에 같이 놀러 가지 않을래?

- Input: Today I have an important meeting.
- Output: Today(2025-02-04) I have an important meeting.

- Input: I wish to become a billionare someday!!
- Output: N/A
</example>
"""
```

```python
## You can change the temperature, but the model must remain as OPENAI_MODELS.GPT4O_MINI.
## Run your results

for test_query in test_list_korean + test_list_english + test_list_no_date:
    all_messages = [{"role": "system", "content": SYSTEM_PROMPT}]
    all_messages.append({"role": "user", "content": test_query})

    response = openai_client.chat.completions.create(
        messages=all_messages, model=OPENAI_MODELS.GPT4O_MINI, temperature=0.7,
    )
    print(test_query)
    print(response.choices[0].message.content)
    print("\n")
```
